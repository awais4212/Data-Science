```python
## Vacumm cleaner Task


def vacumm(room):
    for i in range(len(room)):
        for j in range(len(room[i])):
            if room[i][j] != 0:
               print(f"Cleaning: {room[i][j]}") 
               room[i][j] = 0    
            else:
               print("Already clean")
            
    for row in room:
        print(row)




matrix = [[1,2,3],
          [4,0,6],
          [7,9,8]
]



vacumm(matrix)

    
```
